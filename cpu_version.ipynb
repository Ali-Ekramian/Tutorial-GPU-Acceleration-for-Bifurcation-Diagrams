## CPU code

```python
import time
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.dpi'] = 500

r_values = np.linspace(2.8, 4.0, 50000)
n_iterations = 2000
n_last = 500

t0 = time.time()

r_plot = []
x_plot = []

for r in r_values:
    x = 0.5
    history = []
    for i in range(n_iterations):
        x = r * x * (1 - x)
        history.append(x)
    r_plot.extend([r] * n_last)
    x_plot.extend(history[-n_last:])
    if r%10000==0:
      print("r = ", r)

t_cpu = time.time() - t0
print(f"CPU runtime: {t_cpu:.3f} s")

plt.figure(figsize=(12, 6))
plt.plot(r_plot, x_plot, ',k', markersize=0.05, alpha=0.05)
plt.xlabel("r")
plt.ylabel("x (long-term)")
plt.title(f"Bifurcation diagram (CPU, runtime = {t_cpu:.2f} s)")
plt.show()
```
